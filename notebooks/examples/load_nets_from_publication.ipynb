```python
# Extract the zip file
data_dir = Path("./data")
data_dir.mkdir(parents=True, exist_ok=True)

zip_file_path = Path("./Allostery_PRX_suppl_zenodo.zip")

with zipfile.ZipFile(zip_file_path, "r") as zip_ref:
    zip_ref.extractall(data_dir)
```

```python
# Load a specific network from file dowloaded from Zenodo
file_path = Path("./data/networks/3_1_1_antisymmetric_99.yaml")
net = Network.load_from_file(file_path)
```

```python
# load all networks from the directory
networks = list((data_dir / "networks").glob("*.yaml"))
proteins = list((data_dir / "proteins").glob("*.yaml"))

nets = {}
for net_file in networks + proteins:
    net = Network.load_from_file(net_file)
    nets[net.identifier] = net
```
